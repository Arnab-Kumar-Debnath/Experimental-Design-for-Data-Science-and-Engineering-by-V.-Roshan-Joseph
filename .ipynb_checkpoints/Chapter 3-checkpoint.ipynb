```python
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

plt.style.use("ggplot")
data1=np.linspace(0,1,10)
data2=[0.5+0.5*np.cos(np.pi*((i-0.5)/10)) for i in range(1,11)]
datas=[data1, data2]
R_matrix=np.zeros([len(data2), len(data2)])
```

# RMSE for equispaced and Chevyshevs points

```python
def R(x, theta=0.1):
    return np.exp(-((x**2)/(theta**2)))

def r(x, data):
    x_R=x-data
    return R(x_R)

def s(x, R_matrix,  data):
    return np.sqrt(1-r(x, data).T@np.linalg.inv(R_matrix)@r(x, data))
```

```python
x=np.linspace(0,1,1000)
for data in datas:
    for i in range(len(data)):
        for j in range(len(data)):
            R_matrix[i,j]=R(data[i]-data[j])

    s_=[]
    for i in x:
        s_.append(s(i, R_matrix, data))
    fig, ax = plt.subplots()
    ax.plot(s_)
```

## IMSE Calculation

```python
def Area(y1, y2, del_x):
    return (1/2)*(y1+y2)*del_x
```

```python
#def Point_IMSE()
```

```python
IMSE=[]
for data in datas:
    r_matrix=[]
    R_matrix=[]
    for i in x:
        r_matrix.append(r(np.array([i]), data))
    for j in data:
        R_matrix.append(r(np.array(j), data))
    r_matrix=np.array(r_matrix)
    R_matrix=np.array(R_matrix)
    inverse=np.linalg.inv(R_matrix)
    ys=np.diag(r_matrix@r_matrix.T)
    areas=[]
    for i in range(len(ys)-1):
        areas.append(Area(ys[i], ys[i+1],x[1]-x[0]))
    IMSE.append(1-np.trace(np.sum(areas)*inverse))

    
```
